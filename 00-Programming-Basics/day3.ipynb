```python
#Boolean Values
a=3
b=6
print(a<b)
print(a>b)
print(bool("Hello"))
print(bool(15))
```

```python
#Python Operators
#Arithmetic Operators

print(2+3)
print(3-4)
print(4*5)
print(4%2)
print(6/2)
```

```python
#Assignment Operators
x=5
print(x)
b+=-13
print(b)
x -= 3
print(x)
x/=2
print(x)
x*=2
print(x)
```

```python
#Comparison Operators
a=10
b=20
print(a==b)
print(a<b)
print(a>b)
print(a<=b)
print(a>=b)
print(a!=b)
```

```python
#Logical Operators
a=10
print(a==10 and a>=9)
print(a==10 or a>=9)
print(not(a==10))
```

```python
#Membership Operators
a=[10,20,30,40]
print(10 in a)
print(50 not in a)
```
