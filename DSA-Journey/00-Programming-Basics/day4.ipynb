```python
#Loop Through a List
thislist = ["apple", "banana", "cherry"]
for i in range(len(thislist)):
  print(thislist[i])
```

```python
#Using a While Loop
l1=["apple","banana","mango"]
i=0
while i <(len(l1)):
    print(l1[i])
    i=i+1
```
